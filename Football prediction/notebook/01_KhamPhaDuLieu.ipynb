```python
df = pd.read_csv(r"C:\Users\ADMIN\Desktop\data\final_dataset.csv", index_col=0)
df.columns = df.columns.str.strip()
```

```python
numeric_df = df.select_dtypes(include=['float64', 'int64'])
corr_matrix = numeric_df.corr()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=False, cmap='RdYlGn')
plt.title("Bảng tương quan giữa các biến")
plt.show()
```

```python
print("Top 3 biến tương quan cao nhất với FTHG:")
print(corr_matrix['FTHG'].sort_values(ascending=False)[1:4])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
top_3_manual = ['DiffPts', 'HTGD', 'HTP']
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[top_3_manual], palette="YlOrRd")
plt.title('Kiểm tra Outliers cho DiffPts, HTGD và HTP')
plt.show()
```

## Outliers Detection

Boxplot cho các biến HTP, DiffPts và HTGD cho thấy có xuất hiện một số outliers nằm ngoài vùng phân bố chính của dữ liệu.

### 1. HTP

- Một số giá trị điểm số rất cao xuất hiện ở phía trên boxplot.
- Điều này phản ánh các đội có thành tích vượt trội trong mùa giải.

### 2. DiffPts

- Có các giá trị chênh lệch điểm khá lớn giữa hai đội.
- Các outliers này cho thấy sự khác biệt rõ rệt về sức mạnh đội bóng trong một số trận đấu.

### 3. HTGD

- Xuất hiện cả outliers âm và dương.
- Điều này là hợp lý vì một số đội có hiệu số bàn thắng rất cao hoặc rất thấp.

### Kết luận

- Các outliers được phát hiện chủ yếu phản ánh đặc điểm thực tế của bóng đá thay vì lỗi dữ liệu.
- Vì vậy chưa cần loại bỏ các giá trị này ở bước hiện tại.
- Tuy nhiên cần tiếp tục theo dõi trong quá trình preprocessing và modeling để đảm bảo chúng không ảnh hưởng tiêu cực đến hiệu suất mô hình.

## Categorical Data Analysis

### 1. FTR (Full Time Result)

- Cột FTR gồm 3 giá trị:
  - H: Home Win
  - A: Away Win
  - D: Draw

- Dữ liệu được phân bố tương đối hợp lý.
- Số trận thắng sân nhà thường cao hơn thắng sân khách và hòa, phản ánh lợi thế sân nhà trong bóng đá.

---

### 2. HomeTeam và AwayTeam

- Dataset chứa nhiều đội bóng khác nhau.
- Sau khi kiểm tra các giá trị unique, chưa phát hiện lỗi không đồng nhất hoặc trùng nghĩa trong tên đội bóng.
- Không xuất hiện các trường hợp khác biệt cách viết như viết tắt hoặc sai chính tả.

---

### Kết luận

- Các biến categorical trong dataset nhìn chung sạch và ổn định.
- Không phát hiện missing values hoặc lỗi text nghiêm trọng.
- Dữ liệu phù hợp để thực hiện encoding trong bước preprocessing cho Machine Learning.
